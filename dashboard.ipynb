```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv("hotel_bookings.csv")
print(df)
```

```python
hotel=df["hotel"].value_counts()
print(hotel)

month=df["arrival_date_month"].value_counts()
print(month)
print()

segment=df["market_segment"].value_counts()
print(segment)


# co


waiting=df["days_in_waiting_list"].value_counts()
print(waiting)
```

```python
hotel=df["hotel"].value_counts()
print(hotel)
fig,ax=plt.subplots(2,3,figsize=(18,10))  #figsize=(width,height)
fig.suptitle("hotel booking dashboard",fontsize=20)
hotel.plot(x="hotel",kind="bar",ax=ax[0,0])
ax[0,0].set_title("hotel booking")

df.groupby("hotel")["is_canceled"].sum().plot(kind="bar",ax=ax[0,1])
ax[0,1].set_title("hotel booking cancelled")

# df.groupby("hotel")["is_canceled"].mean().plot(kind="bar",ax=ax[0,2])

month.plot(kind="line",ax=ax[0,2],marker="^",color="red")
ax[0,2].set_title("month booking")


segment.plot(kind="bar",ax=ax[1,0],color=["r","b","g","y","pink","orange","pink","green"])
ax[1,0].set_title("market segment")






waiting.plot(kind="line",ax=ax[1,2],marker="^",color="yellow")
ax[1,2].set_title("waiting")


plt.tight_layout()  #will arrange the name of axis properly without overlapping
plt.show()
```
